```python
import requests
from requests.auth import HTTPBasicAuth
import urllib3
import json
import pandas as pd

# 경고 끄기
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

def load_api(url, save_data=False):  
    # 기본 설정
    username = 'csdtpcp'       # 아이디
    password = 'dtpcp!@'       # 비밀번호
    # 요청 보내기
    response = requests.get(url, auth=HTTPBasicAuth(username, password), verify=False)

    # 응답 확인
    if response.status_code == 200:
        data = response.json()  # JSON 형식으로 파싱
        if save_data:
            # JSON 파일로 저장
            with open('output.json', 'w', encoding='utf-8') as f:
                json.dump(data, f, ensure_ascii=False, indent=4)
    else:
        print(f'접근 실패: 상태 코드 {response.status_code}')
        return False       
    return data
```

```python
import matplotlib.pyplot as plt
import math

# 도시 좌표 (원래 스케일 y축=1)
cities = {
    'A': (0, 4),
    'B': (5, 0),
    'C': (4, 4),
    'D': (5, 5),
    'E': (2, 0)
}

# y축 2배 스케일링된 좌표 생성
cities_scaled = {k: (x, 2 * y) for k, (x, y) in cities.items()}

# 최적 경로 순서
tour_orig   = ['A', 'C', 'D', 'B', 'E', 'A']
tour_scaled = ['A', 'D', 'C', 'B', 'E', 'A']

def tour_length(tour, coords):
    total = 0
    for u, v in zip(tour, tour[1:]):
        x1, y1 = coords[u]
        x2, y2 = coords[v]
        total += math.hypot(x2 - x1, y2 - y1)
    return total

# 목적함수 값 계산
len_orig_on_orig   = tour_length(tour_orig,   cities)
len_scaled_on_orig = tour_length(tour_scaled, cities)
len_orig_on_scaled   = tour_length(tour_orig,   cities_scaled)
len_scaled_on_scaled = tour_length(tour_scaled, cities_scaled)

# 시각화
fig, axes = plt.subplots(1, 2, figsize=(12, 6))
settings = [
    (cities, tour_orig,   tour_scaled, len_orig_on_orig,   len_scaled_on_orig,   'Original'),
    (cities_scaled, tour_scaled, tour_orig,   len_scaled_on_scaled, len_orig_on_scaled, 'Scaled')
]

for ax, (coords, best_tour, alt_tour, best_len, alt_len, title) in zip(axes, settings):
    # 최적 경로 (실선)
    xs = [coords[c][0] for c in best_tour]
    ys = [coords[c][1] for c in best_tour]
    ax.plot(xs, ys, '-o')
    # 대안 경로 (대시선)
    # xs = [coords[c][0] for c in alt_tour]
    # ys = [coords[c][1] for c in alt_tour]
    # ax.plot(xs, ys, '--o')

    for city, (x, y) in coords.items():
        ax.text(x, y, city, fontsize=12, va='bottom', ha='right')

    ax.set_xlim(0, 6)
    ax.set_ylim(0, 11)
    ax.set_aspect('equal', 'box')
    ax.set_title(title)
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    
    ax.text(0.05, 0.95,
            f'Best len: {best_len:.3f}\nOther len: {alt_len:.3f}',
            transform=ax.transAxes, va='top')

plt.tight_layout()
plt.savefig('TSP.png',dpi=1000)
plt.show()
```
